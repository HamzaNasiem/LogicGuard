# Notebook 1 — Knowledge Base Exploration

Interactive exploration of the LogicGuard Knowledge Base.
Visualize graph structure, check coverage, trace BFS paths.

Run: `jupyter notebook notebooks/01_kb_exploration.ipynb`

```python
import sys
sys.path.insert(0, '../src')

from logicguard.kb.loader import KnowledgeBase
from logicguard.kb.validator import BFSValidator
import networkx as nx
import matplotlib.pyplot as plt

kb = KnowledgeBase('../data/knowledge_bases/knowledge_base_extended.json')
print('KB Stats:', kb.stats)
```

```python
# Visualize taxonomy subgraph (mammal branch)
mammals = ['dog', 'cat', 'whale', 'dolphin', 'lion', 'canine', 'feline', 'cetacean', 'mammal', 'animal', 'living_thing']
subgraph = kb.G_T.subgraph(mammals)

plt.figure(figsize=(12, 8))
pos = nx.spring_layout(subgraph, seed=42)
nx.draw(subgraph, pos, with_labels=True, node_color='lightblue', 
        node_size=2000, font_size=10, arrows=True, arrowsize=20)
plt.title('LogicGuard Taxonomy Graph — Mammal Branch')
plt.tight_layout()
plt.savefig('../docs/figures/kb_mammal_subgraph.png', dpi=150)
plt.show()
```

```python
# Interactive BFS path tracer
validator = BFSValidator(kb)

queries = [
    ('dog',    'mammal'),
    ('spider', 'insect'),
    ('whale',  'animal'),
    ('square', 'rectangle'),
]

for subject, predicate in queries:
    result, state, path = validator.validate_taxonomic(subject, predicate)
    path_str = ' → '.join(path) if path else 'No path'
    print(f'{subject:12} IS-A {predicate:12} | {state.value:8} | {str(result):5} | {path_str}')
```
